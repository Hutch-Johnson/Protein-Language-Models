```python
keys = list(esm_dict.keys())

for i in range(len(keys)):
    key = keys[i]
    if 'llr_matrix' in esm_dict[key].keys():
        x = esm_dict[key]['DMS']
        y = esm_dict[key]['llr_matrix']
        seq = esm_dict[key]['sequence']
        
        sp_val, p_val = spearman_ignore_nan(x, y)
        esm_dict[key]['spearman_DMS'] = (sp_val,p_val)
```

```python
i = 25

x = esm_dict[keys[i]]['DMS']
y = esm_dict[keys[i]]['llr_matrix']

sp_val, p_val = esm_dict[keys[i]]['spearman_DMS']

mask = ~np.isnan(x) & ~np.isnan(y)
x_clean = x[mask]
y_clean = y[mask]


# Scatter plot
plt.figure(figsize=(6, 6))
plt.scatter(x_clean, y_clean, color='blue', alpha=0.7, s=5)
plt.xlabel("DMS")
plt.ylabel("LLR")
plt.title(f"DMS vs LLR for {keys[i]}")
plt.grid(False)
ax = plt.gca()
ax.text(
    0.01, 0.99,                
    f"Spearman = {round(sp_val,5)}",
    transform=plt.gca().transAxes,
    fontsize=12,
    verticalalignment='top',
    horizontalalignment='left',
)
ax.spines['right'].set_visible(False)
ax.spines['top'].set_visible(False)


plt.show()
```

```python
indices_to_plot = range(20)  
n_plots = len(indices_to_plot)

# Determine grid size (e.g., 4 rows x 5 columns)
n_rows = 4
n_cols = 5

fig, axes = plt.subplots(n_rows, n_cols, figsize=(20, 12))
axes = axes.flatten()  # flatten for easy indexing

for idx, i in enumerate(indices_to_plot):
    if 'llr_matrix' in esm_dict[keys[i]].keys():
        ax = axes[idx]
        
        x = esm_dict[keys[i]]['DMS']
        y = esm_dict[keys[i]]['llr_matrix']
        seq = esm_dict[keys[i]]['sequence']
        
        sp_val, p_val = esm_dict[keys[i]]['spearman_DMS']
        
        mask = ~np.isnan(x) & ~np.isnan(y)
        x_clean = x[mask]
        y_clean = y[mask]
        
        if sp_val > 0.5:
            ax.scatter(x_clean, y_clean, color='green', alpha=0.5, s=5)
        else:
            ax.scatter(x_clean, y_clean, color='steelblue', alpha=0.5, s=5)
        
        ax.set_title(f'{i}: {keys[i]} (ρ={sp_val:.2f}) len = {len(seq)}', fontsize=10)
        ax.grid(False)
    

# Hide any unused subplots (if grid > n_plots)
for k in range(n_plots, len(axes)):
    axes[k].axis('off')

fig.suptitle("DMS vs LLR for 20 entries", fontsize=16)
plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```
