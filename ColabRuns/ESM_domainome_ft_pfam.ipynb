```python
#@title Loss function
def listwise_ranking_loss(predicts, targets):
    """
    ListMLE loss for ranking tasks.

    Args:
        predicts: Predicted scores
        targets: Target scores

    Returns:
        Mean loss value
    """
    indices = targets.sort(descending=True, dim=-1).indices
    predicts = torch.gather(predicts, dim=1, index=indices)

    cumsums = predicts.exp().flip(dims=[1]).cumsum(dim=1).flip(dims=[1])
    loss = torch.log(cumsums + 1e-10) - predicts

    return loss.mean()
```

```python
#@title Finetuning across all domains (single global model)


# ── Constants ─────────────────────────────────────────────────────────────────

AA_ORDER        = ['L', 'A', 'G', 'V', 'S', 'E', 'R', 'T', 'I', 'D',
                   'P', 'K', 'Q', 'N', 'F', 'Y', 'M', 'H', 'W', 'C']
MODEL_NAME      = "facebook/esm2_t33_650M_UR50D"
OUTPUT_DIR      = "finetuned_models_v2"
MAX_LEN         = 1022
NUM_SAMPLES     = 20
MAX_EPOCHS      = 20
WEIGHT_DECAY    = 1e-2
LORA_DROPOUT    = 0.1
RANDOM_SEED     = 42

# ── Sweep Grid ────────────────────────────────────────────────────────────────

LR_VALUES     = [1e-4, 5e-4, 1e-3, 5e-3]
LORA_R_VALUES = [8, 16, 64, 128]

os.makedirs(OUTPUT_DIR, exist_ok=True)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# ── GCS Helpers ───────────────────────────────────────────────────────────────

from google.cloud import storage

_gcs_client = storage.Client(project=GCS_PROJECT) if GCS_PROJECT else storage.Client()
_gcs_bucket = _gcs_client.bucket(GCS_BUCKET)


def gcs_list_blobs(prefix):
    return list(_gcs_client.list_blobs(GCS_BUCKET, prefix=prefix))


def gcs_download_pickle_gz(blob):
    """Download a gzipped pickle blob and return the unpickled object."""
    data = blob.download_as_bytes()
    with gzip.GzipFile(fileobj=io.BytesIO(data), mode="rb") as gz:
        return pickle.load(gz)


def fitness_matrix_to_rows(fitness, dom_seq, position_offset=0):
    """Convert a [n_pos × 20] fitness matrix into a list of variant dicts
    matching the schema the rest of the pipeline expects.

    `position_offset` is the 0-indexed start of `dom_seq` inside the full
    protein the model is fed. Positions in the returned rows are 1-indexed
    against that full protein (i.e. `position_offset + i + 1`)."""
    rows = []
    for i, wt_aa in enumerate(dom_seq):
        if i >= len(fitness):
            break
        row = fitness[i]
        for j, mut_aa in enumerate(FITNESS_AA_ORDER):
            if j >= len(row):
                continue
            val = row[j]
            if val is None:
                continue
            try:
                fval = float(val)
            except (TypeError, ValueError):
                continue
            if np.isnan(fval):
                continue
            if mut_aa == wt_aa:
                continue
            rows.append({
                "position":           position_offset + i + 1,
                "wt_aa":              wt_aa,
                "mut_aa":             mut_aa,
                "normalized_fitness": fval,
            })
    return rows


def gcs_upload_bytes(blob_path, data, content_type="application/octet-stream"):
    blob = _gcs_bucket.blob(blob_path)
    blob.upload_from_string(data, content_type=content_type)


def gcs_upload_json(blob_path, obj):
    gcs_upload_bytes(blob_path, json.dumps(obj, indent=2, default=str),
                     content_type="application/json")


def gcs_upload_file(blob_path, local_path, content_type="application/octet-stream"):
    blob = _gcs_bucket.blob(blob_path)
    blob.upload_from_filename(local_path, content_type=content_type)


# ── Helper Functions ──────────────────────────────────────────────────────────

def set_seed(seed):
    torch.manual_seed(seed)
    np.random.seed(seed)
    random.seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)


def truncate_to_domain(sequence, domain_start, domain_end, max_len=MAX_LEN):
    seq_len    = len(sequence)
    domain_mid = (domain_start + domain_end) // 2
    half       = max_len // 2
    start      = max(0, domain_mid - 1 - half)
    end        = start + max_len
    if end > seq_len:
        end   = seq_len
        start = max(0, end - max_len)
    return sequence[start:end], start


def get_LLR_scores(sequence, model, tokenizer, device):
    seq_list  = list(sequence)
    tokenized = tokenizer(sequence, padding=False, truncation=True,
                          max_length=1024, return_tensors="pt")
    input_ids = tokenized["input_ids"].to(device)

    logits    = model(input_ids).logits
    log_probs = torch.log_softmax(logits, dim=-1)
    wt_logits = log_probs[:, 1:-1, :].squeeze(0)

    wt_logits_df = pd.DataFrame(
        wt_logits[:, 4:24].cpu().detach().numpy(),
        columns=AA_ORDER,
        index=[f"{aa} {i+1}" for i, aa in enumerate(seq_list)]
    ).T

    wt_norm = np.diag(wt_logits_df.loc[[c.split(" ")[0] for c in wt_logits_df.columns]])
    LLR     = wt_logits_df - wt_norm
    return LLR, wt_logits, seq_list


def lookup_llr_scores(rows, LLR, seq_offset):
    """Score a list of variant dicts against a precomputed LLR table."""
    predicted = []
    for row in rows:
        if (row.get("position") is None or row.get("wt_aa") is None or
            row.get("mut_aa") is None or
            (isinstance(row.get("position"), float) and np.isnan(row["position"]))):
            predicted.append(np.nan)
            continue
        col = f"{row['wt_aa']} {int(row['position']) - seq_offset}"
        mut = row["mut_aa"]
        try:
            predicted.append(LLR.loc[mut, col])
        except KeyError:
            predicted.append(np.nan)
    return predicted


def compute_global_loss_from_samples(global_pool, domains, num_samples,
                                     model, tokenizer, device):
    """Sample `num_samples` variants across all domains, run forward passes
    grouped by sequence, and return `(loss, sampled_domain_ids)`.

    `sampled_domain_ids` is the list of domain ids drawn for this batch
    (one entry per sampled variant; duplicates preserved so callers can
    see per-domain frequency)."""
    valid_pool = [r for r in global_pool
                  if r.get("normalized_fitness") is not None
                  and not (isinstance(r["normalized_fitness"], float)
                           and np.isnan(r["normalized_fitness"]))
                  and r.get("position") is not None
                  and r.get("wt_aa") is not None
                  and r.get("mut_aa") is not None]
    if len(valid_pool) < 2:
        return None, []

    sample = random.sample(valid_pool, min(num_samples, len(valid_pool)))
    sampled_domain_ids = [r["domain_id"] for r in sample]

    # Group sampled variants by domain so each unique sequence is forwarded once.
    by_domain = {}
    for row in sample:
        by_domain.setdefault(row["domain_id"], []).append(row)

    pred_llr = []
    exp_fit  = []

    for domain_id, rows in by_domain.items():
        sequence   = domains[domain_id]["sequence"]
        seq_offset = domains[domain_id]["seq_offset"]

        tokenized = tokenizer(sequence, padding=False, truncation=True,
                              max_length=1024, return_tensors="pt")
        input_ids = tokenized["input_ids"].to(device)
        logits    = model(input_ids).logits
        log_probs = torch.log_softmax(logits, dim=-1)
        wt_logits = log_probs[:, 1:-1, :].squeeze(0)

        for row in rows:
            pos    = int(row["position"]) - seq_offset - 1
            mut_aa = row["mut_aa"]
            wt_aa  = row["wt_aa"]
            if pos < 0 or pos >= wt_logits.shape[0]:
                continue
            if mut_aa not in AA_ORDER or wt_aa not in AA_ORDER:
                continue
            mut_idx    = AA_ORDER.index(mut_aa)
            wt_idx     = AA_ORDER.index(wt_aa)
            pred_score = wt_logits[pos, 4 + mut_idx] - wt_logits[pos, 4 + wt_idx]
            pred_llr.append(pred_score)
            exp_fit.append(float(row["normalized_fitness"]))

    if len(pred_llr) < 2:
        return None, sampled_domain_ids

    pred_tensor = torch.stack(pred_llr).unsqueeze(0)
    exp_tensor  = torch.tensor(exp_fit, dtype=torch.float32, device=device).unsqueeze(0)
    return listwise_ranking_loss(pred_tensor, exp_tensor), sampled_domain_ids


def get_spearman_for_rows(rows, LLR, seq_offset):
    preds = lookup_llr_scores(rows, LLR, seq_offset)
    valid = [(p, r["normalized_fitness"]) for p, r in zip(preds, rows)
             if not (isinstance(p, float) and np.isnan(p))]
    if not valid:
        return float("nan"), 0
    rho, _ = spearmanr(*zip(*valid))
    return rho, len(valid)


def build_lora_model(lora_r, lora_alpha):
    """Build a fresh ESM2 + LoRA model."""
    model       = EsmForMaskedLM.from_pretrained(MODEL_NAME)
    lora_config = LoraConfig(
        task_type      = "FEATURE_EXTRACTION",
        r              = lora_r,
        lora_alpha     = lora_alpha,
        target_modules = ["query", "key", "value", "output.dense"],
        lora_dropout   = LORA_DROPOUT,
    )
    model = get_peft_model(model, lora_config)
    for name, param in model.named_parameters():
        param.requires_grad = "lora" in name
    return model


def evaluate_on_domains(model, tokenizer, domains, splits, device):
    """Compute spearman ρ per domain × split for the given model."""
    scores = {}
    model.eval()
    with torch.no_grad():
        for domain_id, dinfo in domains.items():
            sequence   = dinfo["sequence"]
            seq_offset = dinfo["seq_offset"]
            LLR, _, _  = get_LLR_scores(sequence, model, tokenizer, device)

            scores[domain_id] = {}
            for split_name in ["full", "val", "test"]:
                rows = splits[domain_id][split_name]
                rho, n_valid = get_spearman_for_rows(rows, LLR, seq_offset)
                scores[domain_id][split_name] = {"spearman": rho, "n": n_valid}
    return scores


def run_global_finetune(global_train, global_val, domains, val_splits, tokenizer,
                        lora_r, learning_rate, seed):
    """Fine-tune one LoRA model on samples drawn across the entire dataset."""
    set_seed(seed)
    model = build_lora_model(lora_r, lora_r * 2)
    model.to(device)

    optimizer = torch.optim.AdamW(
        [p for p in model.parameters() if p.requires_grad],
        lr=learning_rate, weight_decay=WEIGHT_DECAY
    )

    best_val_loss   = float("inf")
    best_val_rho    = -float("inf")
    best_lora_state = None
    train_losses    = []
    val_losses      = []
    val_rhos        = []
    train_sampled_domains = []   # per-epoch list of sampled domain ids (train)
    val_sampled_domains   = []   # per-epoch list of sampled domain ids (val)

    for epoch in range(1, MAX_EPOCHS + 1):

        model.train()
        train_loss, train_sampled = compute_global_loss_from_samples(
            global_train, domains, NUM_SAMPLES, model, tokenizer, device
        )
        train_sampled_domains.append(train_sampled)
        if train_loss is None:
            continue

        optimizer.zero_grad()
        train_loss.backward()
        optimizer.step()
        train_losses.append(train_loss.item())

        model.eval()
        with torch.no_grad():
            val_loss, val_sampled = compute_global_loss_from_samples(
                global_val, domains, NUM_SAMPLES, model, tokenizer, device
            )
        val_sampled_domains.append(val_sampled)
        val_loss_val = val_loss.item() if val_loss is not None else float("nan")

        # Mean per-domain val Spearman (averaged across domains with ≥ 2 valid pairs)
        per_domain_rhos = []
        with torch.no_grad():
            for domain_id, dinfo in domains.items():
                rows = val_splits[domain_id]
                if len(rows) < 2:
                    continue
                LLR, _, _ = get_LLR_scores(dinfo["sequence"], model, tokenizer, device)
                rho, n_valid = get_spearman_for_rows(rows, LLR, dinfo["seq_offset"])
                if n_valid >= 2 and not np.isnan(rho):
                    per_domain_rhos.append(rho)
        val_rho = float(np.mean(per_domain_rhos)) if per_domain_rhos else float("nan")

        val_losses.append(val_loss_val)
        val_rhos.append(val_rho)

        if val_loss_val < best_val_loss or val_rho > best_val_rho:
            best_val_loss = min(val_loss_val, best_val_loss)
            best_val_rho  = max(val_rho, best_val_rho)
            best_lora_state = {k: v.cpu().clone()
                               for k, v in model.state_dict().items()
                               if "lora_" in k}

        if len(val_losses) >= 3:
            loss_worse = val_losses[-1] > val_losses[-2] and val_losses[-2] > val_losses[-3]
            rho_worse  = val_rhos[-1]   < val_rhos[-2]   and val_rhos[-2]   < val_rhos[-3]
            if loss_worse or rho_worse:
                break

    del model
    torch.cuda.empty_cache()

    return (best_lora_state, best_val_loss, best_val_rho,
            train_losses, val_losses, val_rhos,
            train_sampled_domains, val_sampled_domains)


def save_sweep_plots(sweep_records, baseline_mean_val_rho, run_id, run_dir):

    # ── Val Spearman Curves ───────────────────────────────────────────────────

    n_lr   = len(LR_VALUES)
    colors = plt.cm.tab10(np.linspace(0, 1, len(LORA_R_VALUES)))

    fig, axes = plt.subplots(1, n_lr, figsize=(5 * n_lr, 5), dpi=150, sharey=True)
    if n_lr == 1:
        axes = [axes]

    for ax, lr in zip(axes, LR_VALUES):
        for color, lora_r in zip(colors, LORA_R_VALUES):
            record = next((r for r in sweep_records
                           if r["lr"] == lr and r["lora_r"] == lora_r), None)
            if record is None:
                continue
            epochs = list(range(1, len(record["val_rhos"]) + 1))
            ax.plot(epochs, record["val_rhos"], color=color, linewidth=2,
                    marker="o", markersize=4, label=f"r={lora_r}")

        ax.axhline(baseline_mean_val_rho, color="black",
                   linewidth=1, linestyle="--", alpha=0.5, label="Baseline")
        ax.set_title(f"lr = {lr:.0e}", fontsize=13)
        ax.set_xlabel("Epoch", fontsize=12)
        ax.legend(fontsize=10, frameon=False)
        for spine in ["top", "right"]:
            ax.spines[spine].set_visible(False)
        for spine in ["left", "bottom"]:
            ax.spines[spine].set_linewidth(1.2)
        ax.tick_params(width=1.2)

    axes[0].set_ylabel("Mean Val Spearman ρ", fontsize=12)
    plt.suptitle(f"{run_id} — Sweep Val Spearman", fontsize=13, y=1.02)
    plt.tight_layout()
    plt.savefig(os.path.join(run_dir, "sweep_val_spearman.png"),
                dpi=150, bbox_inches="tight")
    plt.close()

    # ── Heatmap ───────────────────────────────────────────────────────────────

    rows = [{"lr": r["lr"], "lora_r": r["lora_r"], "best_val_rho": r["best_val_rho"]}
            for r in sweep_records]
    heatmap_data = pd.DataFrame(rows).pivot(index="lora_r", columns="lr",
                                            values="best_val_rho")

    fig, ax = plt.subplots(figsize=(6, 4), dpi=150)
    im = ax.imshow(heatmap_data.values, cmap="RdYlGn", aspect="auto")
    plt.colorbar(im, ax=ax, label="Best Mean Val Spearman ρ")

    ax.set_xticks(range(len(LR_VALUES)))
    ax.set_xticklabels([f"{lr:.0e}" for lr in heatmap_data.columns], fontsize=11)
    ax.set_yticks(range(len(LORA_R_VALUES)))
    ax.set_yticklabels(heatmap_data.index.astype(str), fontsize=11)
    ax.set_xlabel("Learning Rate", fontsize=13)
    ax.set_ylabel("LoRA Rank", fontsize=13)
    ax.set_title(f"{run_id} — Sweep Heatmap", fontsize=13)

    for i in range(len(LORA_R_VALUES)):
        for j in range(len(LR_VALUES)):
            ax.text(j, i, f"{heatmap_data.values[i, j]:.3f}",
                    ha="center", va="center", fontsize=9, color="black")

    plt.tight_layout()
    plt.savefig(os.path.join(run_dir, "sweep_heatmap.png"),
                dpi=150, bbox_inches="tight")
    plt.close()

    # ── Best Config Curves ────────────────────────────────────────────────────

    best_record = max(sweep_records, key=lambda r: r["best_val_rho"])
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4), dpi=150)

    epochs = list(range(1, len(best_record["train_losses"]) + 1))
    ax1.plot(epochs, best_record["train_losses"], color="#708090", linewidth=2,
             marker="o", markersize=4, label="Train")
    ax1.plot(epochs, best_record["val_losses"], color="#d62728", linewidth=2,
             marker="o", markersize=4, label="Val")
    ax1.set_xlabel("Epoch", fontsize=12)
    ax1.set_ylabel("ListMLE Loss", fontsize=12)
    ax1.set_title(f"Best Config Loss\n(lr={best_record['lr']:.0e}, r={best_record['lora_r']})",
                  fontsize=11)
    ax1.legend(fontsize=10, frameon=False)

    ax2.plot(epochs, best_record["val_rhos"], color="#d62728", linewidth=2,
             marker="o", markersize=4, label="Mean Val Spearman")
    ax2.axhline(baseline_mean_val_rho, color="black",
                linewidth=1, linestyle="--", alpha=0.5, label="Baseline")
    ax2.set_xlabel("Epoch", fontsize=12)
    ax2.set_ylabel("Spearman ρ", fontsize=12)
    ax2.set_title("Best Config Mean Val Spearman", fontsize=11)
    ax2.legend(fontsize=10, frameon=False)

    for ax in [ax1, ax2]:
        for spine in ["top", "right"]:
            ax.spines[spine].set_visible(False)
        for spine in ["left", "bottom"]:
            ax.spines[spine].set_linewidth(1.2)
        ax.tick_params(width=1.2)

    plt.suptitle(f"{run_id} — Best Config Curves", fontsize=13, y=1.02)
    plt.tight_layout()
    plt.savefig(os.path.join(run_dir, "best_config_curves.png"),
                dpi=150, bbox_inches="tight")
    plt.close()


# ── Load Shared Resources from GCS ────────────────────────────────────────────

tokenizer = EsmTokenizer.from_pretrained(MODEL_NAME)

# Full protein dict: {uniprot_id: {"sequence": str, "pfam": str, ...}}
full_protein_dict = {}
if GCS_FULL_PROTEIN_BLOB:
    print(f"Loading full protein dict from gs://{GCS_BUCKET}/{GCS_FULL_PROTEIN_BLOB}")
    full_protein_dict = gcs_download_pickle_gz(
        _gcs_bucket.blob(GCS_FULL_PROTEIN_BLOB)
    )
    print(f"Loaded {len(full_protein_dict)} full protein entries")
else:
    print("No GCS_FULL_PROTEIN_BLOB set — falling back to dom_seq for every domain")

# Single gzipped pickle: {domain_id: {"fitness", "uniprot_id", "dom_seq"}}
print(f"Loading fitness dict from gs://{GCS_BUCKET}/{GCS_DOMAIN_DATA_BLOB}")
fitness_dict = gcs_download_pickle_gz(_gcs_bucket.blob(GCS_DOMAIN_DATA_BLOB))
print(f"Loaded {len(fitness_dict)} domain entries")

domain_data = {}    # domain_id → list of variant dicts (positions vs. full protein)
domain_meta = {}    # domain_id → {"uniprot_id", "dom_seq", "full_seq",
                    #              "dom_start_in_protein"}

for domain_id, d in tqdm(fitness_dict.items(), desc="Building variant rows"):
    if not isinstance(d, dict):
        tqdm.write(f"[SKIP] {domain_id} — value is not a dict")
        continue

    fitness    = d.get("fitness")
    uniprot_id = d.get("uniprot_id", domain_id.split("_")[0])
    dom_seq    = d.get("dom_seq")
    if fitness is None or dom_seq is None:
        tqdm.write(f"[SKIP] {domain_id} — missing fitness or dom_seq")
        continue

    # Try to anchor dom_seq inside the full protein. If absent or not a
    # substring, fall back to dom_seq alone (offset 0).
    full_seq             = None
    dom_start_in_protein = 0
    entry = full_protein_dict.get(uniprot_id)
    if entry:
        candidate = entry.get("sequence") if isinstance(entry, dict) else entry
        if isinstance(candidate, str):
            idx = candidate.find(dom_seq)
            if idx >= 0:
                full_seq             = candidate
                dom_start_in_protein = idx
            else:
                tqdm.write(f"[WARN] {domain_id} — dom_seq not in full protein "
                           f"for {uniprot_id}, using dom_seq")

    rows = fitness_matrix_to_rows(fitness, dom_seq,
                                  position_offset=dom_start_in_protein)
    if len(rows) < 10:
        tqdm.write(f"[SKIP] {domain_id} — too few variants ({len(rows)})")
        continue

    domain_data[domain_id] = rows
    domain_meta[domain_id] = {
        "uniprot_id":           uniprot_id,
        "dom_seq":              dom_seq,
        "full_seq":             full_seq,
        "dom_start_in_protein": dom_start_in_protein,
    }

print(f"Loaded {len(domain_data)} usable domains from GCS")
n_with_full = sum(1 for m in domain_meta.values() if m["full_seq"] is not None)
print(f"  using full protein context: {n_with_full} / {len(domain_meta)}")

# ── Build per-domain sequence + splits ────────────────────────────────────────

domains = {}        # domain_id → {sequence, seq_offset, uniprot_id, ...}
splits  = {}        # domain_id → {full, train, val, test} lists of variant dicts

set_seed(RANDOM_SEED)

for domain_id, rows in domain_data.items():

    meta       = domain_meta[domain_id]
    uniprot_id = meta["uniprot_id"]
    dom_seq    = meta["dom_seq"]
    full_seq   = meta["full_seq"]
    dom_start  = meta["dom_start_in_protein"]

    # Prefer the full protein as model input; restrict to a domain-centred
    # window if it exceeds MAX_LEN, exactly like the original notebook.
    seq_offset = 0
    if full_seq is not None:
        sequence = full_seq
        if len(sequence) > MAX_LEN:
            domain_start_1idx = dom_start + 1
            domain_end_1idx   = dom_start + len(dom_seq)
            sequence, seq_offset = truncate_to_domain(
                sequence, domain_start_1idx, domain_end_1idx
            )
    else:
        # Fallback: feed dom_seq directly (positions are already 1-indexed
        # against it because dom_start_in_protein was 0).
        sequence = dom_seq
        if len(sequence) > MAX_LEN:
            positions    = [r["position"] for r in rows]
            domain_start = min(positions)
            domain_end   = max(positions)
            sequence, seq_offset = truncate_to_domain(
                sequence, domain_start, domain_end
            )

    rows_tagged = [dict(r, domain_id=domain_id) for r in rows]

    rng        = np.random.default_rng(RANDOM_SEED)
    perm       = rng.permutation(len(rows_tagged))
    shuffled   = [rows_tagged[i] for i in perm]

    n        = len(shuffled)
    n_train  = int(0.20 * n)
    n_val    = int(0.40 * n)

    train_rows = shuffled[:n_train]
    val_rows   = shuffled[n_train:n_train + n_val]
    test_rows  = shuffled[n_train + n_val:]

    domains[domain_id] = {
        "uniprot_id":           uniprot_id,
        "sequence":             sequence,
        "seq_offset":           seq_offset,
        "seq_len":              len(sequence),
        "dom_start_in_protein": dom_start,
        "used_full_protein":    full_seq is not None,
        "n_total":              n,
        "n_train":              n_train,
        "n_val":                n_val,
        "n_test":               n - n_train - n_val,
    }
    splits[domain_id] = {
        "full":  shuffled,
        "train": train_rows,
        "val":   val_rows,
        "test":  test_rows,
    }

print(f"Prepared splits for {len(domains)} domains")

# ── Group domains by pfam (pfam id is embedded in domain_id as PF#####) ───────

import re

def _extract_pfam(domain_id):
    m = re.search(r"PF\d+", domain_id)
    return m.group(0) if m else None

pfam_to_domain_ids = {}
for domain_id in domains:
    pfam_id = _extract_pfam(domain_id)
    if pfam_id is None:
        print(f"[WARN] no pfam id found in {domain_id} — skipping")
        continue
    pfam_to_domain_ids.setdefault(pfam_id, []).append(domain_id)

print(f"Found {len(pfam_to_domain_ids)} pfams across "
      f"{sum(len(v) for v in pfam_to_domain_ids.values())} domains")

# ── Baseline (un-finetuned ESM2) per domain ───────────────────────────────────

print("Running baseline evaluation on all domains...")
base_model = EsmForMaskedLM.from_pretrained(MODEL_NAME).to(device)
baseline_per_domain = evaluate_on_domains(base_model, tokenizer, domains, splits, device)
del base_model
torch.cuda.empty_cache()

baseline_val_rhos = [d["val"]["spearman"] for d in baseline_per_domain.values()
                     if not np.isnan(d["val"]["spearman"])]
baseline_mean_val_rho = float(np.mean(baseline_val_rhos)) if baseline_val_rhos else 0.0
print(f"Baseline mean val ρ across domains: {baseline_mean_val_rho:.4f}")

# ── Per-pfam Fine-Tuning (one fresh ESM2 + sweep per pfam) ────────────────────

def _gcs_key(local_path):
    rel = os.path.relpath(local_path, OUTPUT_DIR)
    return GCS_OUTPUT_PREFIX + rel.replace(os.sep, "/")

for pfam_id, pfam_domain_ids in pfam_to_domain_ids.items():

    pfam_domains = {d: domains[d] for d in pfam_domain_ids}
    pfam_splits  = {d: splits[d]  for d in pfam_domain_ids}

    # Pools restricted to this pfam's domains only.
    global_train = [r for d in pfam_splits.values() for r in d["train"]]
    global_val   = [r for d in pfam_splits.values() for r in d["val"]]
    global_test  = [r for d in pfam_splits.values() for r in d["test"]]
    print(f"\n[{pfam_id}] {len(pfam_domains)} domains — pool sizes "
          f"train: {len(global_train)}, val: {len(global_val)}, test: {len(global_test)}")

    val_split_rows = {d: pfam_splits[d]["val"] for d in pfam_domains}

    if len(global_train) < 2 or len(global_val) < 2:
        print(f"[{pfam_id}] Skipping — not enough variants")
        continue

    pfam_baseline_val_rhos = [baseline_per_domain[d]["val"]["spearman"]
                              for d in pfam_domain_ids
                              if not np.isnan(baseline_per_domain[d]["val"]["spearman"])]
    pfam_baseline_mean_val_rho = (float(np.mean(pfam_baseline_val_rhos))
                                  if pfam_baseline_val_rhos else 0.0)

    # ── Hyperparameter Sweep (one model per pfam) ─────────────────────────────

    run_id  = f"global_finetune/{pfam_id}"
    run_dir = os.path.join(OUTPUT_DIR, run_id)
    os.makedirs(run_dir, exist_ok=True)

    sweep_records    = []
    best_sweep_rho   = -float("inf")
    best_sweep_state = None
    best_sweep_cfg   = None

    for lr, lora_r in tqdm(list(product(LR_VALUES, LORA_R_VALUES)), desc=f"Sweep {pfam_id}"):
        (best_lora_state, best_val_loss, best_val_rho,
         train_losses, val_losses, val_rhos,
         train_sampled_domains, val_sampled_domains) = run_global_finetune(
            global_train, global_val, pfam_domains, val_split_rows, tokenizer,
            lora_r=lora_r, learning_rate=lr, seed=RANDOM_SEED
        )

        sweep_records.append({
            "lr":                     lr,
            "lora_r":                 lora_r,
            "lora_alpha":             lora_r * 2,
            "best_val_loss":          best_val_loss,
            "best_val_rho":           best_val_rho,
            "epochs_ran":             len(train_losses),
            "train_losses":           train_losses,
            "val_losses":             val_losses,
            "val_rhos":               val_rhos,
            "train_sampled_domains":  train_sampled_domains,
            "val_sampled_domains":    val_sampled_domains,
        })

        if best_val_rho > best_sweep_rho:
            best_sweep_rho   = best_val_rho
            best_sweep_state = best_lora_state
            best_sweep_cfg   = {"lr": lr, "lora_r": lora_r, "lora_alpha": lora_r * 2}

    if best_sweep_state is None:
        print(f"[{pfam_id}] No successful sweep run — skipping save")
        continue

    # ── Save Sweep Plots ──────────────────────────────────────────────────────

    save_sweep_plots(sweep_records, pfam_baseline_mean_val_rho, run_id, run_dir)

    # ── Evaluate Best Model on All Splits Per-Domain ──────────────────────────

    best_model = build_lora_model(best_sweep_cfg["lora_r"], best_sweep_cfg["lora_alpha"])
    best_model.load_state_dict(best_sweep_state, strict=False)
    best_model.to(device)

    ft_per_domain = evaluate_on_domains(best_model, tokenizer, pfam_domains, pfam_splits, device)

    # ── Save Outputs Locally + Upload to GCS as Dictionaries ──────────────────

    # 1. LoRA-only weights
    lora_path = os.path.join(run_dir, "best_model_lora.pth")
    torch.save(best_sweep_state, lora_path)

    # 2. Full sweep curves + summary as JSON dict
    sweep_curves_path = os.path.join(run_dir, "sweep_curves.json")
    with open(sweep_curves_path, "w") as f:
        json.dump(sweep_records, f, indent=2, default=str)

    # 3. Per-domain master log (dict of dicts)
    master_log = {}
    for domain_id, dinfo in pfam_domains.items():
        base = baseline_per_domain[domain_id]
        ft   = ft_per_domain[domain_id]
        master_log[domain_id] = {
            "domain_id":         domain_id,
            "uniprot_id":        dinfo["uniprot_id"],
            "pfam_id":           pfam_id,
            "n_variants":        dinfo["n_total"],
            "baseline_val_rho":  base["val"]["spearman"],
            "baseline_test_rho": base["test"]["spearman"],
            "ft_val_rho":        ft["val"]["spearman"],
            "ft_test_rho":       ft["test"]["spearman"],
            "delta_val_rho":     ft["val"]["spearman"] - base["val"]["spearman"],
            "delta_test_rho":    ft["test"]["spearman"] - base["test"]["spearman"],
        }

    # 4. Aggregate metadata dict
    delta_val_rhos  = [v["delta_val_rho"]  for v in master_log.values()
                       if not np.isnan(v["delta_val_rho"])]
    delta_test_rhos = [v["delta_test_rho"] for v in master_log.values()
                       if not np.isnan(v["delta_test_rho"])]

    metadata = {
        "run_id":                run_id,
        "pfam_id":               pfam_id,
        "n_domains":             len(pfam_domains),
        "global_pool_sizes": {
            "train": len(global_train),
            "val":   len(global_val),
            "test":  len(global_test),
        },
        "random_seed":           RANDOM_SEED,
        "best_config":           best_sweep_cfg,
        "baseline_mean_val_rho": pfam_baseline_mean_val_rho,
        "mean_delta_val_rho":    float(np.mean(delta_val_rhos))  if delta_val_rhos  else float("nan"),
        "mean_delta_test_rho":   float(np.mean(delta_test_rhos)) if delta_test_rhos else float("nan"),
        "baseline_per_domain":   {d: baseline_per_domain[d] for d in pfam_domain_ids},
        "finetuned_per_domain":  ft_per_domain,
        "domains":               pfam_domains,
        "sweep_grid": {
            "lr_values":     LR_VALUES,
            "lora_r_values": LORA_R_VALUES,
            "max_epochs":    MAX_EPOCHS,
            "num_samples":   NUM_SAMPLES,
            "weight_decay":  WEIGHT_DECAY,
            "lora_dropout":  LORA_DROPOUT,
        },
    }

    metadata_path = os.path.join(run_dir, "metadata.json")
    with open(metadata_path, "w") as f:
        json.dump(metadata, f, indent=2, default=str)

    master_log_path = os.path.join(run_dir, "master_log.json")
    with open(master_log_path, "w") as f:
        json.dump(master_log, f, indent=2, default=str)

    # ── Upload Everything to GCS ──────────────────────────────────────────────

    print(f"[{pfam_id}] Uploading results to GCS...")
    gcs_upload_file(_gcs_key(lora_path),         lora_path)
    gcs_upload_file(_gcs_key(sweep_curves_path), sweep_curves_path, "application/json")
    gcs_upload_file(_gcs_key(metadata_path),     metadata_path,     "application/json")
    gcs_upload_file(_gcs_key(master_log_path),   master_log_path,   "application/json")
    for fname in ["sweep_val_spearman.png", "sweep_heatmap.png", "best_config_curves.png"]:
        fpath = os.path.join(run_dir, fname)
        if os.path.exists(fpath):
            gcs_upload_file(_gcs_key(fpath), fpath, "image/png")

    del best_model
    torch.cuda.empty_cache()

    print(f"[{pfam_id}] Done — best lr={best_sweep_cfg['lr']:.0e}, r={best_sweep_cfg['lora_r']}, "
          f"Δ val ρ={metadata['mean_delta_val_rho']:.4f}, Δ test ρ={metadata['mean_delta_test_rho']:.4f}")

# ── Final Summary ─────────────────────────────────────────────────────────────

print(f"\n── All pfams done ──")
print(f"Pfams processed: {len(pfam_to_domain_ids)}")
print(f"Local results: {OUTPUT_DIR}/")
print(f"GCS results:   gs://{GCS_BUCKET}/{GCS_OUTPUT_PREFIX}")


# ── HOW TO RELOAD A SAVED MODEL ───────────────────────────────────────────────
#
# The run saves only LoRA delta weights (~10-50MB) in best_model_lora.pth
# (locally and at gs://<bucket>/<prefix>/global_finetune/best_model_lora.pth).
# To reload for inference:
#
#   import json, torch
#   from transformers import EsmForMaskedLM, EsmTokenizer
#   from peft import LoraConfig, get_peft_model
#
#   run_dir = "finetuned_models_v2/global_finetune"
#
#   # 1. Load metadata to recover the saved config
#   with open(f"{run_dir}/metadata.json") as f:
#       metadata = json.load(f)
#   cfg = metadata["best_config"]   # has lora_r, lora_alpha
#
#   # 2. Rebuild the base + LoRA architecture (same as training)
#   model = build_lora_model(cfg["lora_r"], cfg["lora_alpha"])
#
#   # 3. Load LoRA weights — strict=False fills base ESM2 weights from pretrained
#   lora_weights = torch.load(f"{run_dir}/best_model_lora.pth")
#   model.load_state_dict(lora_weights, strict=False)
#
#   # 4. Ready for inference
#   model.to(device)
#   model.eval()
#   tokenizer = EsmTokenizer.from_pretrained("facebook/esm2_t33_650M_UR50D")
#   LLR, _, _ = get_LLR_scores(sequence, model, tokenizer, device)
#
# ─────────────────────────────────────────────────────────────────────────────
```
